#### Function to create the graph for other columns as well

```python
def functionToCreateGraph(figsize, values, column_name):
    plt.figure()
    values.plot(figsize=figsize)
    plt.xlabel("years") #the pandas .plot() function automatically uses the index of the DataFrame for the x-axis.
    plt.ylabel(column_name)
    plt.title(f"{column_name} of Hindalco Data")
```

```python
hindalcoData.columns
```

```python
for data in hindalcoData.columns:
    functionToCreateGraph((5,5),hindalcoData[data],data)
```
