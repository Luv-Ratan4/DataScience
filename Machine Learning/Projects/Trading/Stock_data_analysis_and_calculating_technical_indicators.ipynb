#### Importing necessary libraries

```python
import yfinance as yf
import numpy as np
import pandas as pd
import pandas_ta as ta

```

#### Importing the data from Yahoo Finance

```python
data = pd.DataFrame(yf.download('HINDALCO.ns', period='6mo', interval='1h'))
data.reset_index(inplace=True)  #Converting date time into an index
data.tail()
```

#### Cleaning the data

```python
data.isna().sum()   #Checking for null values
```

```python
indexZeroes = data[data['Volume'] == 0].index #Find the index of the data/rows where volume is 0
data.drop(indexZeroes, inplace=True)
data.loc[data["Volume"]==0]
```

#### Loading the Pandas Technical Analysis Library

```python
data['ATR'] = ta.atr(close=data['Close'], high=data['High'], low=data['Low'],length=20)
data['RSI'] = ta.rsi(close=data['Close'])
data['Average'] = ta.midprice( high=data['High'], low=data['Low'], length=1)
data['MA40'] = ta.sma(close=data['Close'], length=40)
data['MA80'] = ta.sma(close=data['Close'], length=80)
data['MA160'] = ta.sma(close=data['Close'], length=160)
data
```
