### Conditional Statements - If/Else

#### Notes
- We can have multiple elif only between if and else

```python
n = int(input("Enter your number:"))
if(n == 0):
    print("The Number is zero")
elif(n == 1 or n== 2):
    print("The number is neither odd nor even")
elif(n%2 == 0):
    print(n,"is even")
else:
    print(n,"is odd")
```

### Loops

#### For-Loop

- In Range, we write 1 more than what we want since the numbering starts from 0

```python
n = int(input("Enter your number: "))
for i in range(n):
    print(i)

for i in range(n):
    print('*'*i)


```

```python
n = 15
for i in range(1,11):       #We have given starting of the range and ending of the range
    print(n,'*',i,'=',i*n)
```

```python
#Taking more than one step at a time
n = 10
for i in range(1,11,2): #Third paramenter is the number of steps
    print(i)
```

```python
n = 10
for i in range(2,21,2):
    print(n,'*',i,'=',i*n)
```

#### Nested For-loops

```python
for i in range(1,7):    # i loop will exceute once for every 7 times the j loop executes.
    for j in range(1,7):
        for k in range(1,7):
            if(i+j+k == 5):
                print(i, j, k)
                print('-'*15)
```

#### While Loop

- For loop is used when know the certain number of iterations that we want to make
- While loop is used when we don't know the certain number of iterations that we will need to make
- While loop is used when we want to end the loop when the certain condition is met.

```python
n = 30
i = 1
s = 1

while(n>i):
   
    print(s)
    i += 1
    s += i
```

```python
# Program to reverse a number

number = 5484848
n = number
reversedNumber = 0
while(number>0):
    temp = number%10
    reversedNumber = reversedNumber * 10 + temp
    number //= 10

print(reversedNumber)
print(n)



```
