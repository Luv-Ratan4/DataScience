## Different types of inbuilt data structures in python

- List (Most Important for data science)
- Dictionary (Also really important for data science)
- Set and Tuple

### Lists

- List is a collection of different types of data that we can put together.
- Represented by a square bracket, just like an array
- We can have any data type of our choice and we can have multiple types of data together in a list
- Every item is seperated by a comma

#### Major list functions

```python
lst = [15,56,252,414,1]
lst2 = ["Luv", "Ratan", "Arya", "Vinod", "Ratan"]
lst3 = [True, False, True, True, False, True, False, False]
lst4 = [15.25,626.15,65.21,878.25]
lst5 = ["Luv", 45415, True, 15.15]
print(type(lst),type(lst2),type(lst3),type(lst4),type(lst5))
print(lst,lst2,lst3,lst4,lst5, sep="   ")
```

##### Accessing elements of a list

```python
print(lst[0])   #Accessing the first element of the list which is at index zero since index starts with zero in a list
print(lst2[2])
print(lst3[-1])
print(type(lst5[0])) #When we return only one item, the returned type is the data type of that element
```

##### Modifying the values in a list

```python
lst[2] = 'New Value'
print(lst)
```

##### Slicing a list

```python
print(lst[1:3]) #Returns the value from the list starting from the given index till one less than the other index
print(type(lst[1:3]))   #Returned type is also a list if we are returning more than one element
print(lst[:])   #Returns the complete string
```

##### Reversing a list

```python
print(lst[::-1])
print(lst[::-2])    #Reverse using two steps at a time
```

##### Appending a list
- Adding another item or a complete new list to an existing list

```python
print(lst)
lst.append("New Item appended")
print(lst)
lst.append(lst2) #This will not add individual list items seperately, but will become a nested list
print(lst)
```

##### Removing an item from the list

```python
print(lst)
lst.remove(lst2)    #It will search if there is list inside the list that have the same items as lst2 and then remove it
print(lst)
lst.remove(414)     #This will remove the given item from the list
print(lst)
```

```python
#Length of list
print(len(lst))
```

##### Sorting a list
- only numeric values can be sorted

```python
print(sorted(lst4))
```

##### Finding the index of the element in the given list

```python
print(lst.index('New Value'))
```

##### Counting the occurence of element in the list

```python
print(lst.count("New Value"))
```

##### Concatenating/extending a list

```python
lst.extend(lst2)
print(lst)  #This will concatenate the two lists together with each items as seperate elements
lst.extend(["One","two","Three"])
print(lst)
```

##### Finding Maximum and minimum

```python
print(min(lst4))    #Prints the minimum of the list
print(max(lst4))    #Prints the maximum of the list
lst4    
```

#### Iterating through a list

- There are two ways to access an element from a list
    - Loop
    - Indexing

##### For-loop to iterating through the elements of the list    | Direct

```python
print(lst)
for i in lst:
    print(i, lst.index(i), lst[lst.index(i)], sep=" ,  ")
```

##### Indexing to iterate through a list | Indexing

```python
print(lst)
for i in range(len(lst)):
    print(lst[i], i, sep=" ,  ")
```

```python
# Accessing elements in reverse

print(lst)
for i in range(len(lst)):
    if(i>=1):
        print(lst[len(lst) - i], len(lst)- i, sep=" ,  ")

print('-'*50)

#OR we can also do 

for i in range(len(lst) -1, -1, -1):
    print(lst[i])
```
