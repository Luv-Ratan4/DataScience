#### Data Types
There are four main types of data -

- Integer - e.g - -4545,-11,-2,0,2,55,11,1100           ------------Numbers
- Float - e.g. - 1.5, 0.0, -1.55, 5.9, 8.85, -9.2       ------------Numbers 
- String - "Hello", "WORLD", 'Boy', 'Girl'              ------------Characters  
- Boolean - True/False
