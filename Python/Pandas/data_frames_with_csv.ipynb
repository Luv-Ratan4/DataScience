### Removing the null values

```python
link = 'https://raw.githubusercontent.com/AshishJangra27/Data-Analysis-with-Python-GFG/main/3.%20Data%20Preprocessing%20-%20Removing%20Null%20Value%20Rows/googleplaystore.csv'

data = pd.read_csv(link)
df = data.copy()
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df.dropna(inplace=True, axis=0) #Drops the null values row wise
df.isnull().sum()
```

#### Replacing the null values

```python
data.isna().sum()
```

```python
missing = round(data['Rating'].mean(),2) #This will give us the average rating and then round it off
print(missing)
data['Rating'] = data['Rating'].fillna(missing)    #This will fill the missing values in the Ratings column
data.isna().sum()
```

```python
# Filling up categorical data

repeatedValues = data['Current Ver'].mode()  #Mode will give us, which value is occuring most number of times
print(repeatedValues, type(repeatedValues))
data['Current Ver'] = data['Current Ver'].fillna(str(repeatedValues))
data.isna().sum()
```
